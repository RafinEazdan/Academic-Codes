```python
import heapq
def ucs(graph, start, goal):
    visited = set()
    priority_queue = [(0, start,[])]
    heapq.heapify(priority_queue)
    while priority_queue:
        cost, node, path = heapq.heappop(priority_queue)
        if node in visited:
            continue
        visited.add(node)
        path = path + [node]
        if node == goal:
            print(f"Goal {goal} found")
            print(f"Path: {path}")
            print(f"Cost: {cost}")
            return
            
        for n,c in graph[node].items():
            if n not in visited:
                heapq.heappush(priority_queue, (cost + c, n, path))
    
    print(f"Goal {goal} not found")
    print(f"Path: {path}")

```

```python
graph = {
    'S': {'A': 1, 'B': 2, 'C': 3},
    'A': {'D': 2,'E':3},
    'B': {'G': 2},
    'C': {'F': 4},
    'E': {'G': 1},
    'G': {},
    'F': {},
    'D': {'H':9},
    'H': {}
}

goal = 'G'
start = 'S'
ucs(graph, start, goal)
```
