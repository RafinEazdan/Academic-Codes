```python
def dfs(graph, start, goal):
    visited = set()
    stack = [start]
    path = []
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            path.append(node)
            if node == goal:
                print(f"Goal {goal} found")
                print(f"Path: {path}")
                return
            
            for n in graph[node]:
                if n not in visited:
                    stack.append(n)
    
    print(f"Goal {goal} not found")
    print(f"Path: {path}")

```

```python
graph = {
    'S': ['A', 'B', 'C'],
    'A': ['S', 'D', 'E'],
    'B': ['S', 'C', 'E'],
    'C': ['S', 'B', 'F'],
    'D': ['A', 'H'],
    'E': ['A', 'B'],
    'F': ['C', 'G'],
    'G': ['F'],
    'H': ['D']
}

goal_node = 'G'
dfs(graph, 'S', goal_node)
```
