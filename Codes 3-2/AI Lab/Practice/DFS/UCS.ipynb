```python
import heapq  # We need this to create a simple priority queue

def ucs(graph, start, goal):
    # List of nodes to check (priority queue)
    queue = [(0, start)]  # (cost, node)
    
    # Dictionary to remember the cost to reach each node
    cost_to_reach = {start: 0}
    
    # Dictionary to remember the path
    came_from = {start: None}
    
    # While there are still nodes to explore
    while queue:
        # Get the node with the lowest cost
        current_cost, current_node = heapq.heappop(queue)
        
        # If we've reached the goal, reconstruct the path
        if current_node == goal:
            path = []
            while current_node is not None:
                path.append(current_node)
                current_node = came_from[current_node]
            path.reverse()  # Path is constructed backwards, so we reverse it
            return path, cost_to_reach[goal]
        
        # Check each neighbor
        for neighbor in graph[current_node]:
            # Every edge has a cost of 1 (uniform cost)
            new_cost = current_cost + 1
            
            # If this new path is cheaper or the neighbor hasn't been visited
            if neighbor not in cost_to_reach or new_cost < cost_to_reach[neighbor]:
                cost_to_reach[neighbor] = new_cost
                heapq.heappush(queue, (new_cost, neighbor))  # Add neighbor to queue
                came_from[neighbor] = current_node  # Remember where we came from
    
    # If no path is found, return None
    return None, None

# Example graph
graph = {
    'S': ['A', 'B', 'C'],
    'A': ['S', 'D', 'E'],
    'B': ['S', 'C', 'E'],
    'C': ['S', 'B', 'F'],
    'D': ['A', 'H'],
    'E': ['A', 'B'],
    'F': ['C', 'G'],
    'G': ['F'],
    'H': ['D']
}

# Running the UCS algorithm
start_node = 'S'
goal_node = 'G'
path, cost = ucs(graph, start_node, goal_node)

# Output the result
if path:
    print(f"Path found: {path}")
    print(f"Total cost: {cost}")
else:
    print("No path found")
```
