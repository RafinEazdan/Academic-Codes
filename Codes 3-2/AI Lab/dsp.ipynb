# Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# Defining the signal and filter and Their Centres

```python
x = [1,2,3]
h = [-1,2,2]
center_x = 0 
center_h = 0 
x_n = []
```

# Convolution Matrix

```python
def create_scaled_matrix(signal_a, signal_b):
    a = np.array(signal_a)
    b = np.array(signal_b)
    b = b.reshape(-1, 1)
    M = b * a
    return M

create_scaled_matrix(x, h)
```

# Convolution using WEZNEY Method

```python
def convolve_signals(a, b):
    N = len(a)
    M = len(b)
    result_length = N + M - 1
    result = [0] * result_length

    for k in range(result_length):
        summation = 0
        terms = []
        for i in range(N):
            j = k - i
            if 0 <= j < M:
                product = a[i] * b[j]
                summation += product
                terms.append(f"{a[i]}*{b[j]}={product}")
        result[k] = summation
    return result

# Convolution
convolved_signal = convolve_signals(x, h)

# Display the result
print("\nConvolved Signal:")
print(convolved_signal)
```
