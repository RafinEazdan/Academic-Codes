# Sobel

```python
# Sobel operator
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("road.bmp", cv2.IMREAD_GRAYSCALE)

Gx = np.array([[-1, -2, -1],
               [ 0,  0,  0],
               [ 1,  2,  1]], dtype=np.float32)

Gy = np.array([[-1,  0,  1],
               [-2,  0,  2],
               [-1,  0,  1]], dtype=np.float32)

gx = cv2.filter2D(img, ddepth=cv2.CV_32F, kernel=Gx)  # Gradient in X
gy = cv2.filter2D(img, ddepth=cv2.CV_32F, kernel=Gy)  # Gradient in Y

gx_vis = cv2.convertScaleAbs(gx)
gy_vis = cv2.convertScaleAbs(gy)

mag = cv2.magnitude(gx, gy)  # sqrt(gx^2 + gy^2)

plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1), plt.imshow(img, cmap="gray"), plt.title("Original"), plt.axis("off")
plt.subplot(2, 3, 2), plt.imshow(gx_vis, cmap="gray"), plt.title("Sobel X"), plt.axis("off")
plt.subplot(2, 3, 3), plt.imshow(gy_vis, cmap="gray"), plt.title("Sobel Y"), plt.axis("off")
plt.subplot(2, 3, 4), plt.imshow(mag, cmap="gray"), plt.title("Gradient Magnitude"), plt.axis("off")

plt.tight_layout()
plt.show()
```

# Prewitt

```python
img = cv2.imread("horse-dog.jpg", cv2.IMREAD_GRAYSCALE)

Gx = np.array([[-1, -1, -1],
               [ 0,  0,  0],
               [ 1,  1,  1]], dtype=np.float32)

Gy = np.array([[-1,  0,  1],
               [-1,  0,  1],
               [-1,  0,  1]], dtype=np.float32)

gx = cv2.filter2D(img, ddepth=cv2.CV_32F, kernel=Gx)  # Gradient in X
gy = cv2.filter2D(img, ddepth=cv2.CV_32F, kernel=Gy)  # Gradient in Y

gx_vis = cv2.convertScaleAbs(gx)
gy_vis = cv2.convertScaleAbs(gy)

mag = cv2.magnitude(gx, gy)  

plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1), plt.imshow(img, cmap="gray"), plt.title("Original"), plt.axis("off")
plt.subplot(2, 3, 2), plt.imshow(gx_vis, cmap="gray"), plt.title("Sobel X"), plt.axis("off")
plt.subplot(2, 3, 3), plt.imshow(gy_vis, cmap="gray"), plt.title("Sobel Y"), plt.axis("off")
plt.subplot(2, 3, 4), plt.imshow(mag, cmap="gray"), plt.title("Gradient Magnitude"), plt.axis("off")

plt.tight_layout()
plt.show()
```
