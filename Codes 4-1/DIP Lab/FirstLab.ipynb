# 4

```python
img1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow("BGR to RGB", img1)
cv2.imshow("BGR to Gray", img2)
cv2.waitKey()
```

# 5. Resize

```python
resized = cv2.resize(img, (300,300))
cv2.imshow("Resized to 300x300", resized)
cv2.waitKey()
```

# 6. Crop

```python
cropped = img[50:400, 50:500]
cv2.imshow("Cropped Image",cropped)
cv2.waitKey()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image in BGR format (default in OpenCV)
bgr_img = img

# Convert BGR to RGB using OpenCV's built-in function
rgb_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)

# Plot side-by-side using matplotlib
plt.figure(figsize=(10, 5))

# Show BGR image directly (will appear bluish)
plt.subplot(1, 2, 1)
plt.imshow(bgr_img)
plt.title("BGR Image (Wrong Colors)")
plt.axis("off")

# Show correctly converted RGB image
plt.subplot(1, 2, 2)
plt.imshow(rgb_img)
plt.title("RGB Image (Correct Colors)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
