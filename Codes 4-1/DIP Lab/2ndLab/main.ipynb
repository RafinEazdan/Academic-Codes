```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
img = cv2.imread("/Users/eazdanmostafarafin/Downloads/lena_gray.jpg")
```

```python
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# Resize by 1.5x
import cv2

# Read the image
img = cv2.imread('image.jpg')

# Resize by 1.5x
resized = cv2.resize(img, None, fx=1.5, fy=1.5, interpolation=cv2.INTER_LINEAR)

# Show results
cv2.imshow("Original", img)
cv2.imshow("Resized (1.5x)", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Image Negative

```python
# Image Negative
rows, cols = img.shape
negative_img = np.zeros_like(img)
for i in range(rows):
    for j in range(cols):
        negative_img[i, j] = 255 - img[i, j]
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(negative_img, cmap='gray')
plt.title('Negative')
plt.axis('off')
```

# Log Transformation

```python
# Log Transformation
max_value = np.max(img)
min_value = np.min(img)
print(f"Max pixel value: {max_value}, Min pixel value: {min_value}")
c = 255 / np.log(1 + np.max(img))
log_transformed = c * np.log(1 + img)
log_transformed = np.array(log_transformed, dtype=np.uint8)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(log_transformed, cmap='gray')
plt.title('Log Transformed Image')
plt.axis('off')
```

# Gamma Correction

```python
# Gamma Correction
gamma = 2
gamma_corrected = np.array(255 * (img / 255) ** gamma, dtype=np.uint8)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gamma_corrected, cmap='gray')
plt.title('Gamma Corrected Image')
plt.axis('off')
```

# Min-Max Normalization

```python
# Min-Max Normalization
r_min = np.min(img)
r_max = np.max(img)
print(f"Range of pixel values: {r_min} to {r_max}")
min_max_scaled = ((img - r_min) / (r_max - r_min)) * 255
min_max_scaled = np.array(min_max_scaled, dtype=np.uint8)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(min_max_scaled, cmap='gray')
plt.title('Min-Max Normalized Image')
plt.axis('off')
```

# Thresholding

```python
# Thresholding
T = 127
threshold_img = np.where(img > T, 255, 0).astype(np.uint8)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(threshold_img, cmap='gray')
plt.title('Thresholded Image')
plt.axis('off')
```

# Intensity Slicing

```python
# Define slicing range
lower = 100
upper = 150

# 1. Intensity Slicing Without Background Suppression
sliced_without_bg = img.copy()
sliced_without_bg[(img >= lower) & (img <= upper)] = 255

# 2. Intensity Slicing With Background Suppression
sliced_with_bg = np.zeros_like(img)
sliced_with_bg[(img >= lower) & (img <= upper)] = 255

# Display all images using subplots
plt.figure(figsize=(10, 6))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(sliced_without_bg, cmap='gray')
plt.title('Sliced (No BG Suppression)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sliced_with_bg, cmap='gray')
plt.title('Sliced (With BG Suppression)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
