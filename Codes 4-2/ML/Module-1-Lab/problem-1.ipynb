```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("work.csv")
df.head()
```

```python
df.shape
```

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
print(X)
```

```python
print(y)
```

```python
lr = LinearRegression()
model = lr.fit(X,y)
y_hat = model.predict(X)
```

```python
y_hat
```

```python
mae = mean_absolute_error(y, y_hat)
mse = mean_squared_error(y, y_hat)
r2 = r2_score(y, y_hat)
```

```python
print(f"Mean Absolute Error: {mae}")
print(f"Mean Squared Error: {mse}")
print(f"R2 Scored: {r2}")
```

```python
# r2 manually
upper = np.sum(square(y-y_hat))
lower = np.sum(square(y - np.mean(y)))

r2 = 1 - (upper/lower)

print(r2)
```

```python
real_data = int(input("Enter Weekly Study Hour: "))
y_pred_real = model.predict([[real_data]])
print(f"Your Productivity score (estimate): {y_pred_real}")
```

```python
b = model.coef_
a = model.intercept_
print(b,a)
```

```python
x_points = []
y_points = []
for i in range(0,10):
    x = i
    y_new = a+b*i
    x_points.append(x)
    y_points.append(y_new)
plt.figure(figsize=(14,9))
plt.scatter(X,y)
plt.plot(x_points, y_points)
```
