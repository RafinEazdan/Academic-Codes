```python
# Python program for DDA line generation

from matplotlib import pyplot as plt

# DDA Function for line generation


def DDA(x0, y0, x1, y1):

    # find absolute differences
    dx = abs(x0 - x1)
    dy = abs(y0 - y1)

    # find maximum difference
    steps = max(dx, dy)

    # calculate the increment in x and y
    xinc = dx/steps
    yinc = dy/steps

    # start with 1st point
    x = float(x0)
    y = float(y0)

    # make a list for coordinates
    x_coorinates = []
    y_coorinates = []

    for i in range(steps):
        # append the x,y coordinates in respective list
        x_coorinates.append(x)
        y_coorinates.append(y)

        # increment the values
        x = x + xinc
        y = y + yinc

    # plot the line with coordinates list
    plt.plot(x_coorinates, y_coorinates, marker="o",
             markersize=1, markerfacecolor="green")
    plt.show()


# Driver code
if __name__ == "__main__":

    # coordinates of 1st point
    x0, y0 = 20, 20

    # coordinates of 2nd point
    x1, y1 = 60, 50

    # Function call
    DDA(x0, y0, x1, y1)

    # This code is contributed by 111arpit1
```

```python
# Python program for DDA line drawing algorithm
import matplotlib.pyplot as plt

def DDA_Line(x1, y1, x2, y2):
    """
    DDA (Digital Differential Analyzer) Line Drawing Algorithm
    Steps:
    1. Input two endpoints (x1,y1) and (x2,y2)
    2. Calculate dx and dy
    3. Calculate slope m = dy/dx
    4. Set initial point (x1,y1)
    5. Loop through x-coordinates, calculate y using y = y1 + m(x - x1)
    6. Plot each pixel
    7. Repeat until endpoint (x2,y2) is reached
    """
    
    # Step 2: Calculate the difference between coordinates
    dx = x2 - x1
    dy = y2 - y1
    
    # Step 3: Calculate the slope
    if dx != 0:
        m = dy / dx
    else:
        m = None  # Vertical line case
    
    # Step 4: Set the initial point
    x = x1
    y = y1
    
    # Lists to store pixel coordinates
    x_points = []
    y_points = []
    
    # Step 5-7: Loop through x-coordinates and calculate y
    if dx != 0:
        step = 1 if x2 > x1 else -1
        for x in range(x1, x2 + step, step):
            # Calculate y using the equation y = y1 + m(x - x1)
            y = y1 + m * (x - x1)
            # Step 6: Plot the pixel
            x_points.append(x)
            y_points.append(round(y))
    else:
        # Handle vertical line
        step = 1 if y2 > y1 else -1
        for y in range(y1, y2 + step, step):
            x_points.append(x1)
            y_points.append(y)
    
    # Plotting
    plt.figure(figsize=(8, 6))
    plt.plot(x_points, y_points, marker="o", markersize=4, markerfacecolor="green", 
             linestyle='-', linewidth=1, color='green', label='DDA Line')
    plt.title(f"DDA Line Drawing Algorithm: ({x1},{y1}) to ({x2},{y2})")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.gca().set_aspect("equal")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

# DDA Line Example
x1, y1 = 2, 3
x2, y2 = 10, 7
DDA_Line(x1, y1, x2, y2)
```

```python
# Python program for Bresenham's Line Drawing Algorithm
import matplotlib.pyplot as plt

def bresenham_line(x1, y1, x2, y2):
    """
    Bresenham's Line Drawing Algorithm
    Uses only integer arithmetic and incremental error calculation
    """
    # Calculate differences
    dx = abs(x2 - x1)
    dy = abs(y2 - y1)
    
    # Determine direction of line
    x_step = 1 if x1 < x2 else -1
    y_step = 1 if y1 < y2 else -1
    
    # Initialize starting point
    x = x1
    y = y1
    
    # Lists to store pixel coordinates
    x_points = [x]
    y_points = [y]
    
    # Case 1: Slope < 1 (more horizontal)
    if dx > dy:
        # Initial decision parameter
        p = 2 * dy - dx
        
        # Loop through all x coordinates
        for _ in range(dx):
            x += x_step
            
            if p < 0:
                # Choose E (East)
                p = p + 2 * dy
            else:
                # Choose NE (North-East)
                y += y_step
                p = p + 2 * dy - 2 * dx
            
            x_points.append(x)
            y_points.append(y)
    
    # Case 2: Slope >= 1 (more vertical)
    else:
        # Initial decision parameter
        p = 2 * dx - dy
        
        # Loop through all y coordinates
        for _ in range(dy):
            y += y_step
            
            if p < 0:
                # Choose N (North)
                p = p + 2 * dx
            else:
                # Choose NE (North-East)
                x += x_step
                p = p + 2 * dx - 2 * dy
            
            x_points.append(x)
            y_points.append(y)
    
    # Plotting
    plt.figure(figsize=(8, 6))
    plt.plot(x_points, y_points, marker="o", markersize=4, markerfacecolor="blue", 
             linestyle='-', linewidth=1, color='blue', label='Bresenham Line')
    plt.title(f"Bresenham Line Drawing Algorithm: ({x1},{y1}) to ({x2},{y2})")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.gca().set_aspect("equal")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

# Bresenham Line Example
x1, y1 = 2, 3
x2, y2 = 10, 7
bresenham_line(x1, y1, x2, y2)
```
