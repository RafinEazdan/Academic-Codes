# Mid-point Circle

```python
import matplotlib.pyplot as plt

def plot_circle_points(cx, cy, x, y, points):
    """Add 8 symmetric points of a circle"""
    points.extend([
        (cx + x, cy + y),
        (cx - x, cy + y),
        (cx + x, cy - y),
        (cx - x, cy - y),
        (cx + y, cy + x),
        (cx - y, cy + x),
        (cx + y, cy - x),
        (cx - y, cy - x)
    ])

def midpoint_circle(cx, cy, r):
    x = 0
    y = r
    p = 1 - r    

    points = []
    plot_circle_points(cx, cy, x, y, points)

    while x < y:
        x += 1

        if p < 0:
            #  right 
            p = p + 2*x + 1
        else:
            #  right-down 
            y -= 1
            p = p + 2*x + 1 - 2*y

        plot_circle_points(cx, cy, x, y, points)

    return points

# --------- ---------
cx, cy, r = 0, 0, 20
circle_points = midpoint_circle(cx, cy, r)

x_vals = [p[0] for p in circle_points]
y_vals = [p[1] for p in circle_points]

plt.scatter(x_vals, y_vals, s=10)
plt.gca().set_aspect('equal')
plt.title("Midpoint Circle Algorithm")
plt.grid(True)
plt.show()
```

# Bresenham Circle

```python
import matplotlib.pyplot as plt

def plot_circle_points(cx, cy, x, y, points):
    """Plot all 8 symmetric points"""
    points.extend([
        (cx + x, cy + y),
        (cx - x, cy + y),
        (cx + x, cy - y),
        (cx - x, cy - y),
        (cx + y, cy + x),
        (cx - y, cy + x),
        (cx + y, cy - x),
        (cx - y, cy - x)
    ])

def bresenham_circle(cx, cy, r):
    x = 0
    y = r
    d = 3 - 2 * r   

    points = []
    plot_circle_points(cx, cy, x, y, points)

    while x <= y:
        x += 1

        if d < 0:
            # right
            d = d + 4 * x + 6
        else:
            # right-down
            y -= 1
            d = d + 4 * (x - y) + 10

        plot_circle_points(cx, cy, x, y, points)

    return points

# -------- --------
cx, cy, r = 0, 0, 20
circle_points = bresenham_circle(cx, cy, r)

x_vals = [p[0] for p in circle_points]
y_vals = [p[1] for p in circle_points]

plt.scatter(x_vals, y_vals, s=10)
plt.gca().set_aspect('equal')
plt.title("Bresenham Circle Algorithm")
plt.grid(True)
plt.show()
```
