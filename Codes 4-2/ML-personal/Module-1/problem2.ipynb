```python
# Problem 2: Sugar Level Prediction - Model Evaluation
from sklearn.metrics import fbeta_score
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Real and predicted sugar values
real = np.array([2.4, 4.6, 7.3, 4.1, 3.3, 6.8, 5.7])
predicted = np.array([2.2, 3.9, 6.7, 5.2, 3.2, 6.1, 6.2])

# i) MSE (Mean Squared Error)
mse_sugar = mean_squared_error(real, predicted)

# ii) MAE (Mean Absolute Error)
mae_sugar = mean_absolute_error(real, predicted)

# iii) RMSE (Root Mean Squared Error)
rmse_sugar = np.sqrt(mse_sugar)

# iv) F2-Score (convert to binary classification: threshold at median)
threshold = np.median(real)
real_binary = (real >= threshold).astype(int)
predicted_binary = (predicted >= threshold).astype(int)
f2_sugar = fbeta_score(real_binary, predicted_binary, beta=2)

print("\nResults:")
print(f"i)   MSE:      {mse_sugar:.4f}")
print(f"ii)  MAE:      {mae_sugar:.4f}")
print(f"iii) RMSE:     {rmse_sugar:.4f}")
print(f"iv)  F2-Score: {f2_sugar:.4f}")
```
